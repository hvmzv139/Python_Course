Lists are created using squares brackets:


```python
# create a list
thislist = ['apple' , 'banana' , 'cherry']
print(thislist)
```

## **list items**

list items are odered, changeable , and allow duplicate values.

list items are indexed, the first item has index [0], the second item has index [1]etc.



## **Ordered**
When we say that lists are ordered, it means that the items have a defined order, and the order will not change.

if you add new items to a list , the new items will be placed at the end of this list .

ps : there are some list methods , that will change the order , but in general : the order of the items will not change

## **Changeable**
the list is changeable, meaning that we can change, add , and remove items in a list after it has been created.

## **Allow Duplicates**

since lists are indexed , lists can have items with the same valu:



```python
# list allow duplicates value:
thislist = ['apple' , 'cherry' , 'cherry']
thislist

#list length , to determine how many items a list has, use the len() function :
print(len(thislist))

#data types : list items can be any of data types :
#string  , int , boolean data types.
list1 = ['apple' , 'banana' , 'cherry']
list2 = [1 , 5 , 7 , 9 , 3]
list3 = [True , False , False]
print(list1)
print(list2)
print(list3)
# a list contain different data types  , a list with strings , integers , and boolean values
list4 = ['cherry', 15 , True , 20.5]
print(list4)

#form Python perspective  , lists are defined as objects with data types 'list':

mylist = ['apple' , 'banana' , 'cherry']
print(type(mylist))
```

## **The list() Constructor**

it is also possible to use the list() constructor when creating a new list

iterable: An object that you can loop over, one element at a time.

ps : Iterable = something you can iterate over (loop through) one element at a time.


In Python, an object is basically a value stored in memory that Python can work with.


we mean that list() can take something like a string, tuple, list, set, or range and iterate through its elements.

```python
# using the list() constructor to make a list

thislist = list(('apple' , 'banana' , 'cherry'))
print(thislist)
```

## **Python Collections (Arrays)**

there are four collection data types in the python programming language

**list** : is a collection which is ordered and changeable.Allows duplicate members.

**tuples** : is a collection which is ordered and unchangeable.allows duplicate members.

**set** : is a collection  which is unordered, unchangeable , and unidexed , no duplicate members.

**dictionary** : is a collection which is ordered , and changeable , no duplicate members.


set items are unchangeable , but u can add /remove items whenever you like.

as of python version 3.7, dictionaries are ordered  , in python 3.6 and earlier , dictionaries are unordered.

when chosing a collection type , it is useful to understand the properties of that type.Choosing the right type for a particular data set could mean retention of meaning , and , it could mean an increase in effciency or security.
