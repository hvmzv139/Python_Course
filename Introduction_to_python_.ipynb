## **Looping Using List Comprehension**

Looping Using List Comprehension offers the shortest syntax for looping through lists:



```python
# a short list for loop that will print all items in a list
thislist = ["apple", "banana", "cherry"]
[print(x) for x in thislist]
```

A list comprehension is a short way to create a new list by looping through an iterable. In [print(x) for x in thislist], Python loops through each item and calls print(x), so the items are displayed. However, print() only displays a value and does not return a value, so it returns None. Therefore, the resulting list contains None for each item: [None, None, None].

## **List Comprehension**
List comprehension offers a shorter syntax when you want to create a new list based on the values of an existing list.


Based on a list of fruits, you want a new list, containing only the fruits with the letter "a" in the name.

Without list comprehension you will have to write a for statement with a conditional test inside:



```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = []

for x in fruits:
  if 'a' in x :
    newlist.append(x)

print(newlist)
```

BUT With list comprehension you can do all that with only one line of code:



```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = [x for x in fruits if 'a' in x]
newlist
```

**The syntax**


newlist = [expression for item in iterable if condition == True]

the return value is a new list  , leave the last list unchanged

## **Condition**

the condition is like a filter that only accepts the items that evaluate to true.



```python
# only accept items that are not "apple"

fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = [x for x in fruits if x != "apple"]
newlist
#The condition is optional and can be omitted:


newlist = [x for x in fruits]
newlist
```

The condition if x != "apple"  will return True for all elements other than "apple", making the new list contain all fruits except "apple".

## **Expression**

the expression , is the current item in the iteration , but it is also outcome, which you can manipulate before it ends up like a list item in the new list

```python
#Set the values in the new list to upper case:
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist=[x.upper() for x in fruits]
newlist
```

You can set the outcome to whatever you like:



```python
#Set all values in the new list to 'hello':
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = ['hello' for x in fruits]
newlist
```

The expression can also contain conditions, not like a filter, but as a way to manipulate the outcome:



```python
#Return "orange" instead of "banana":
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]

newlist = [x if x != "banana" else "orange" for x in fruits]
newlist
```

The expression in the example above says:

"Return the item if it is not banana, if it is banana return orange".

## **Sort List Alphanumerically**

lists objects have a sort() method that will sort the list alphanumerically , accending by default.

```python
#Sort the list alphabetically:

thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]
thislist.sort()
thislist
```

```python
#Sort the list numerically:

thislist = [100, 50, 65, 82, 23]
thislist.sort()
thislist
```

```python
#To sort descending, use the keyword argument reverse = True:
thislist = [100, 50, 65, 82, 23]
thislist.sort(reverse=True)
thislist
```

## **Customize Sort Function**

You can also customize your own function by using the keyword argument key = function.

The function will return a number that will be used to sort the list (the lowest number first):



```python
# Sort the list based on how close the number is to 50:
def myfunc(n):
  return abs(n - 50)

thislist = [100, 50, 65, 82, 23]
thislist.sort(key = myfunc)
print(thislist)
```

## **Case Insensitive Sort**

By default the sort() method is case sensitive, resulting in all capital letters being sorted before lower case letters:



```python
#Case sensitive sorting can give an unexpected result:
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.sort()
print(thislist)

```

ps : Luckily we can use built-in functions as key functions when sorting a list.So if you want a case-insensitive sort function, use str.lower as a key function:



```python
# Perform a case-insensitive sort of the list:

thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.sort(key = str.lower)
print(thislist)
```

## **Join Two Lists**

There are several ways to join, or concatenate, two or more lists in Python.

One of the easiest ways are by using the + operator.

```python
list1 = ["a", "b", "c"]
list2 = [1, 2, 3]
list3 = list1 + list2
list3
```

Another way to join two lists is by appending all the items from list2 into list1, one by one:



```python
list1 = ["a", "b", "c"]
list2 = [1, 2, 3]
for x in list2 :
  list1.append(x)
list1
```

Or you can use the extend() method, where the purpose is to add elements from one list to another list:



```python
list1 = ["a", "b" , "c"]
list2 = [1, 2, 3]
list1.extend(list2)
print(list1)
```
