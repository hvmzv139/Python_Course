## Multiline Comments
Python does not really have a syntax for multiline comments

```python
#this is a comment
#written in
#more than just one line
print("Hello worls !")
```

```python
"""
this is a comment
written in
more just one line

"""
print("Hello worls !")
```

##**Python - Global Variables**

### **Global Variables:**
Variables that are created outside of a function

Global variables can be used by everyone , both inside of functions and outside


```python
# Create a variable outside a functio and use it inside
Number="139"
def myFunc():
  print("hvmzv",Number)

myFunc()
```

ps : if u create a variable with the same name inside a function, this variable wil be local , and can only be used inside the function. The global variable with the same name will remain as it was , global and with the original value

**Exemple**

```python
des1 = "nice"

def MyFunction():
  des2 = "wonderful"
  print("the weather is",des2)

MyFunction()
print("the weather is",des1)

```

### **The global Keyword**
when we creat a variable inside a fucntion, that variable is local , and only , and can only be used that inside function   

to create a global variable inside a function , use the *global* keyword

```python
# to make a local variable global
def newfucntion():
  global x
  x = "i am a global variable"

newfucntion()
print(x)
```

ps:also use the global keyword if u want to change a value of a  global variable inside a function


```python
# to change  a value of a  global variable inside a function
X = "i am a global variable"
def newfucntion():
  global X
  X="i am a local variable"
  print(X)


newfucntion()
print(X)
```
